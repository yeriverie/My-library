```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler, StandardScaler, LabelEncoder

# ==========================================
# STEP 1: CREATE SAMPLE DATASET
# ==========================================
print("--- 1. Creating Sample Dataset ---")
data = {
    'Employee_ID': [101, 102, 103, 104, 105, 106, 107, 108],
    'Age': [25, 30, np.nan, 35, 40, np.nan, 28, 50],
    'Salary': [50000, 60000, 52000, np.nan, 90000, 62000, 58000, 110000],
    'Department': ['IT', 'HR', 'IT', 'Finance', np.nan, 'HR', 'IT', 'Finance'],
    'Purchased_Course': ['Yes', 'No', 'No', 'Yes', 'Yes', 'No', 'Yes', 'No'],
    'Constant_Column': ['Same_Value'] * 8  # Used to demonstrate feature selection
}

df = pd.DataFrame(data)
print("Original Raw Data:")
print(df)
print("\n" + "="*50 + "\n")

# ==========================================
# STEP 2: EXPLORATORY DATA ANALYSIS (EDA)
# ==========================================
print("--- 2. Exploratory Data Analysis (EDA) ---")
print("Dataset Info:")
df.info()

print("\nMissing Values Count:")
print(df.isnull().sum())

print("\nSummary Statistics:")
print(df.describe(include='all'))
print("\n" + "="*50 + "\n")

# ==========================================
# STEP 3: HANDLING MISSING VALUES
# ==========================================
print("--- 3. Handling Missing Values ---")
# Numerical columns: Fill with Median
df['Age'].fillna(df['Age'].median(), inplace=True)
df['Salary'].fillna(df['Salary'].median(), inplace=True)

# Categorical column: Fill with Mode
mode_dept = df['Department'].mode()[0]
df['Department'].fillna(mode_dept, inplace=True)

print("Data after filling missing values:")
print(df)
print("\n" + "="*50 + "\n")

# ==========================================
# STEP 4: ENCODING CATEGORICAL VARIABLES
# ==========================================
print("--- 4. Encoding Categorical Variables ---")
# Binary Encoding for Target Variable (Purchased_Course)
le = LabelEncoder()
df['Purchased_Course'] = le.fit_transform(df['Purchased_Course'])

# One-Hot Encoding for Multi-class Feature (Department)
df = pd.get_dummies(df, columns=['Department'], prefix='Dept', drop_first=False)

print("Data after Categorical Encoding:")
print(df)
print("\n" + "="*50 + "\n")

# ==========================================
# STEP 5: FEATURE SELECTION
# ==========================================
print("--- 5. Feature Selection ---")
# Drop identifiers (Employee_ID) and zero-variance columns (Constant_Column)
df_selected = df.drop(columns=['Employee_ID', 'Constant_Column'])

print("Data after removing redundant features:")
print(df_selected.head())
print("\n" + "="*50 + "\n")

# ==========================================
# STEP 6: FEATURE NORMALIZATION & SCALING
# ==========================================
print("--- 6. Normalization / Scaling ---")
# Min-Max Scaling for Age and Salary
scaler = MinMaxScaler()
df_selected[['Age', 'Salary']] = scaler.fit_transform(df_selected[['Age', 'Salary']])

print("Final Cleaned & Preprocessed Dataset:")
print(df_selected)

# Save output to CSV
df_selected.to_csv("cleaned_dataset.csv", index=False)
print("\nCleaned dataset saved as 'cleaned_dataset.csv'")
```
